```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/bangalore_tech_salaries.csv')
print("Dataset Loaded Successfully! ")

# Data inspection
print(f"Shape of the data: {df.shape} \n")
print(f"Columns of the data: {df.columns} \n")
print(f"information: {df.info()} \n")
print(f"Null values: {df.isnull().sum()} \n")
print(f"Sample data: {df.head()} \n")
```

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)

df.drop_duplicates(inplace=True)

print("Shape after removing duplicates:", df.shape)
```

```python
df['Current_CTC']=df['Current_CTC'].str.replace('LPA','')
df['Previous_CTC']=df['Previous_CTC'].str.replace('LPA','')

df['Current_CTC']=df['Current_CTC'].str.replace('₹','')
df['Previous_CTC']=df['Previous_CTC'].str.replace('₹','')

print(df['Current_CTC'])
print("\n")
print(df['Previous_CTC'])
```

To convert the strings into numbers


```python
# Convert Current_CTC and Previous_CTC into float (LPA)

def clean_ctc(value):

    if pd.isna(value):
        return np.nan

    value = str(value)
    value = value.replace(",", "")
    value = value.strip()

    try:
        num = float(value)

        # If salary is stored in rupees, convert it to LPA
        if num > 1000:
            num = num / 100000

        return num

    except:
        return np.nan


df["current_ctc"] = df["current_ctc"].apply(clean_ctc)
df["previous_ctc"] = df["previous_ctc"].apply(clean_ctc)

print(df[["current_ctc", "previous_ctc"]].head())
print(df[["current_ctc", "previous_ctc"]].dtypes)
```

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(" ", "_"))

print(df.columns)
```

```python
print(df["role"].unique())
```

```python
# Standardize role names

role_map = {

    "DS": "Data Scientist",
    "Data Science Engineer": "Data Scientist",

    "DA": "Data Analyst",
    "BI Analyst": "Data Analyst",
    "Analytics Engineer": "Data Analyst",

    "BA": "Business Analyst",
    "Business Systems Analyst": "Business Analyst",

    "PM": "Product Manager",
    "Sr PM": "Product Manager",
    "Product Lead": "Product Manager",

    "UI/UX": "UI/UX Designer",
    "Designer": "UI/UX Designer",
    "UI Designer": "UI/UX Designer",
    "UX Designer": "UI/UX Designer",
    "Product Designer": "UI/UX Designer",

    "DevOps": "DevOps Engineer",

    "SRE": "Site Reliability Engineer",
    "Infra Engineer": "Site Reliability Engineer",

    "BE": "SDE Backend",
    "Backend Dev": "SDE Backend",
    "Backend Developer": "SDE Backend",
    "Backend Engineer": "SDE Backend",
    "SDE-Backend": "SDE Backend",

    "FE": "SDE Frontend",
    "Frontend Dev": "SDE Frontend",
    "Frontend Developer": "SDE Frontend",
    "Frontend Engineer": "SDE Frontend",
    "SDE-Frontend": "SDE Frontend",

    "SDE FS": "SDE Full-Stack",
    "FullStack": "SDE Full-Stack",
    "Full Stack Engineer": "SDE Full-Stack",
    "Fullstack Dev": "SDE Full-Stack"

}

df["role_clean"] = df["role"].replace(role_map)

print(df["role_clean"].value_counts())
```

```python
print(df["company_type"].unique())
```

```python
# Mapping company type

company_map = {
    "unicorn": "Unicorn",
    "UNICORN": "Unicorn",

    "mnc": "MNC",

    "mid-size": "Mid-size",
    "MID-SIZE": "Mid-size",

    "early-stage": "Early-stage",
    "EARLY-STAGE": "Early-stage"
}

df["company_type_clean"] = (
    df["company_type"]
    .replace(company_map)
)

print(df["company_type_clean"].value_counts())
```

```python
print(df["education_tier"].unique())
```

```python
# Standardize education tier

education_map = {
    "Tier-1": "Tier 1",
    "Tier-2": "Tier 2",
    "Tier-3": "Tier 3",
    "T1": "Tier 1",
    "T2": "Tier 2",
    "T3": "Tier 3",
    "1": "Tier 1",
    "2": "Tier 2",
    "3": "Tier 3"
}

df["education_tier_clean"] = (
    df["education_tier"]
    .astype(str)
    .replace(education_map)
)

print(df["education_tier_clean"].value_counts())
```

```python
# Fill missing values in Skills and Location

df["skills"] = df["skills"].fillna("Not Available")
df["location"] = df["location"].fillna("Unknown")

# Keeping Previous_CTC as NaN because freshers don't have previous salary

print(df.isnull().sum())
```

```python
print("\nData Types\n")
print(df.dtypes)

print("\nRole Count\n")
print(df["role_clean"].value_counts())

print("\nCompany Type Count\n")
print(df["company_type_clean"].value_counts())

print("\nEducation Tier Count\n")
print(df["education_tier_clean"].value_counts())
```

```python
# Q3.1 - Calculate median, mean, minimum and maximum salary for each role

role_salary = (
    df.groupby("role_clean")["current_ctc"]
    .agg(["median", "mean", "min", "max"])
    .sort_values(by="median", ascending=False)
)
print(role_salary)
print("\n")
highest = role_salary.index[0]
lowest = role_salary.index[-1]

print("\nHighest Paying Role :", highest)
print("Lowest Paying Role :", lowest)
```

### Insight (Q3.1)

The role with the highest median salary is Product Manager, while Data Analyst has the lowest median salary.
The difference between mean and median shows whether some roles have high salary outliers.

```python
# Q3.2 - Compare salary based on experience for SDE Backend

backend = df[df["role_clean"] == "SDE Backend"].copy()

# Create experience bands
bins = [-1, 1, 3, 5, 100]
labels = ["0-1 Years", "2-3 Years", "4-5 Years", "6+ Years"]

backend["exp_band"] = pd.cut(
    backend["years_exp"],
    bins=bins,
    labels=labels
)

# Median salary of each experience band
exp_salary = backend.groupby("exp_band", observed=False)["current_ctc"].median()

print("Median Salary by Experience Band\n")
print(exp_salary)

print("\nGrowth Percentage\n")

previous = None

for band in exp_salary.index:

    salary = exp_salary[band]

    if previous is None:
        print(f"{band} : {salary:.2f} LPA")
    else:
        growth = ((salary - previous) / previous) * 100
        print(f"{band} : {salary:.2f} LPA (+{growth:.2f}%)")

    previous = salary
```

```python
# Q3.4 - Compare SDE Backend salary across company types

backend = df[df["role_clean"] == "SDE Backend"]

company_salary = (
    backend.groupby("company_type_clean")["current_ctc"]
    .median()
)

print(company_salary)

unicorn = company_salary["Unicorn"]
mnc = company_salary["MNC"]

premium = ((unicorn - mnc) / mnc) * 100

print("\nUnicorn Premium over MNC =", round(premium,2),"%")
```

```python
# Q3.5 - Find employees earning below their group's median salary

bins = [0,1,3,5,50]
labels = ["0-1 Years","2-3 Years","4-5 Years","6+ Years"]

df["exp_band"] = pd.cut(
    df["years_exp"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

group_size = (
    df.groupby(
        ["role_clean","company_type_clean","exp_band"]
    )["current_ctc"]
    .transform("count")
)

group_median = (
    df.groupby(
        ["role_clean","company_type_clean","exp_band"]
    )["current_ctc"]
    .transform("median")
)

underpaid = df[group_size >= 10].copy()

underpaid["group_median"] = group_median[group_size >= 10]

underpaid["gap"] = (
    underpaid["current_ctc"] -
    underpaid["group_median"]
)

underpaid = underpaid.sort_values("gap").head(10)

print(
    underpaid[
        [
            "employee_id",
            "role_clean",
            "company_type_clean",
            "years_exp",
            "current_ctc",
            "group_median",
            "gap"
        ]
    ]
)
```

```python
# ==============================
# TASK 4 - FINAL PRINTED REPORT
# ==============================

print("=" * 60)
print("           BANGALORE TECH SALARY DECODER")
print("      Built by Vaishanavi Kashid | The Unlox Academy")
print("=" * 60)

print("Dataset : 1000 Bengaluru Tech Professionals")
print("Period  : 2024 Employment Snapshot")

# ------------------------------------
# Median CTC by Role
# ------------------------------------

print("\n---- MEDIAN CTC BY ROLE (LPA) ----")

for role, salary in role_salary["median"].items():

    bar = "█" * int(salary // 2)

    print(f"{role:<28} {bar:<18} {salary:.2f}")

# ------------------------------------
# SDE Backend Experience
# ------------------------------------

print("\n---- SDE BACKEND CTC BY EXPERIENCE ----")

previous = None

for band in exp_salary.index:

    salary = exp_salary[band]

    if previous is None:

        print(f"{band:<12}: {salary:.2f} LPA")

    else:

        growth = ((salary - previous) / previous) * 100

        print(f"{band:<12}: {salary:.2f} LPA (+{growth:.2f}%)")

    previous = salary

# ------------------------------------
# Skill Premium
# ------------------------------------

print("\n---- SKILL PREMIUM FOR SDEs ----")

skills = ["AWS", "ML", "System Design", "Kubernetes"]

print(f"{'Skill':<18}{'With':<12}{'Without':<12}{'Premium'}")

sde = df[df["role_clean"].isin([
    "SDE Backend",
    "SDE Frontend",
    "SDE Full-Stack"
])]

for skill in skills:

    with_skill = sde[
        sde["skills"].str.contains(skill, case=False, na=False)
    ]["current_ctc"].median()

    without_skill = sde[
        ~sde["skills"].str.contains(skill, case=False, na=False)
    ]["current_ctc"].median()

    premium = ((with_skill - without_skill) / without_skill) * 100

    print(f"{skill:<18}{with_skill:<12.2f}{without_skill:<12.2f}+{premium:.2f}%")

# ------------------------------------
# Company Premium
# ------------------------------------

print("\n---- COMPANY TYPE PREMIUM (SDE Backend) ----")

highest = company_salary.max()

for company, salary in company_salary.items():

    diff = ((highest - salary) / highest) * 100

    if company == company_salary.idxmax():

        print(f"{company:<15}{salary:.2f} LPA  <-- Highest")

    else:

        print(f"{company:<15}{salary:.2f} LPA (-{diff:.2f}%)")

# ------------------------------------
# Top 5 Underpaid Employees
# ------------------------------------

print("\n---- TOP 5 MOST UNDERPAID PROFESSIONALS ----")

top5 = underpaid.head(5)

for _, row in top5.iterrows():

    print(
        f"{row['employee_id']}  "
        f"{row['role_clean']:<18}"
        f"{row['company_type_clean']:<12}"
        f"{row['years_exp']} yrs   "
        f"Gap : {row['gap']:.2f} LPA"
    )

print("=" * 60)
print("                END OF REPORT")
print("=" * 60)
```
